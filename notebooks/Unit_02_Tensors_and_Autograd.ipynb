## 2.3 广播机制(Broadcasting)

广播机制允许不同形状的张量进行算术运算，PyTorch会自动扩展较小的张量以匹配较大张量的形状。

### 广播规则
1. 从尾部维度开始对齐
2. 维度兼容：相等或其中一个为1
3. 自动扩展尺寸为1的维度

```python
print("=" * 60)
print("2.3 广播机制演示")
print("=" * 60)

a = torch.tensor([[1.0, 2.0, 3.0]])
b = torch.tensor([[10.0], [20.0], [30.0]])
print(f"a形状: {a.shape}, a =\n{a}")
print(f"b形状: {b.shape}, b =\n{b}")

result = a + b
print(f"\na + b 结果形状: {result.shape}")
print(f"a + b =\n{result}")

x = torch.randn(3, 1, 4)
y = torch.randn(1, 5, 1)
z = x + y
print(f"\n广播示例2:")
print(f"x形状: {x.shape}, y形状: {y.shape}")
print(f"x + y 结果形状: {z.shape}")
```

## 2.4 张量拼接与分割

```python
print("=" * 60)
print("2.4 张量拼接与分割")
print("=" * 60)

x1 = torch.randn(2, 3)
x2 = torch.randn(2, 3)
x3 = torch.randn(2, 3)

concat_dim0 = torch.cat([x1, x2, x3], dim=0)
print(f"沿dim=0拼接: {concat_dim0.shape}")

concat_dim1 = torch.cat([x1, x2, x3], dim=1)
print(f"沿dim=1拼接: {concat_dim1.shape}")

stacked = torch.stack([x1, x2, x3], dim=0)
print(f"堆叠(stack): {stacked.shape}")

x = torch.randn(6, 4)
chunks = torch.chunk(x, chunks=3, dim=0)
print(f"\n分割为3块: {[c.shape for c in chunks]}")

splits = torch.split(x, split_size_or_sections=2, dim=0)
print(f"按大小2分割: {[s.shape for s in splits]}")
```

## 2.5 自动求导(Autograd)机制

自动求导是PyTorch的核心功能，它能够自动计算张量操作的梯度。

### 关键概念
- **requires_grad**: 标记需要计算梯度的张量
- **grad_fn**: 记录创建张量的操作
- **backward()**: 执行反向传播计算梯度
- **grad**: 存储计算得到的梯度值

```python
print("=" * 60)
print("2.5.1 自动求导基础")
print("=" * 60)

x = torch.tensor(3.0, requires_grad=True)
y = x ** 2 + 2 * x + 1

print(f"x = {x}")
print(f"y = x^2 + 2x + 1 = {y}")
print(f"y.grad_fn = {y.grad_fn}")

y.backward()

print(f"dy/dx = 2x + 2 = {x.grad}")
print(f"在x=3处的梯度: {x.grad.item()}")
```

```python
print("=" * 60)
print("2.5.2 多变量自动求导")
print("=" * 60)

x = torch.tensor([2.0, 3.0], requires_grad=True)
y = x[0] ** 2 + x[1] ** 3

print(f"x = {x}")
print(f"y = x[0]^2 + x[1]^3 = {y}")

y.backward()

print(f"梯度 dy/dx = [2*x[0], 3*x[1]^2] = {x.grad}")
print(f"预期梯度: [4.0, 27.0]")
```

## 2.6 计算图(Computational Graph)

计算图是自动求导的基础，PyTorch使用动态计算图，在每次前向传播时构建。

```python
print("=" * 60)
print("2.6 计算图演示")
print("=" * 60)

a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(3.0, requires_grad=True)

c = a * b
d = a + b
e = c + d

print(f"a = {a}, requires_grad = {a.requires_grad}")
print(f"b = {b}, requires_grad = {b.requires_grad}")
print(f"c = a * b = {c}, grad_fn = {c.grad_fn}")
print(f"d = a + b = {d}, grad_fn = {d.grad_fn}")
print(f"e = c + d = {e}, grad_fn = {e.grad_fn}")

e.backward()

print(f"\n反向传播后:")
print(f"de/da = {a.grad}")
print(f"de/db = {b.grad}")
print(f"理论值: de/da = b+1 = 4.0, de/db = a+1 = 3.0")
```

## 2.7 梯度管理

在实际训练中，需要正确管理梯度以避免梯度累积等问题。

```python
print("=" * 60)
print("2.7.1 梯度清零")
print("=" * 60)

x = torch.tensor(2.0, requires_grad=True)

for i in range(3):
    y = x ** 2
    y.backward()
    print(f"第{i+1}次backward后, x.grad = {x.grad}")

print("\n梯度累积了！需要手动清零。")

x.grad.zero_()
print(f"清零后, x.grad = {x.grad}")
```

```python
print("=" * 60)
print("2.7.2 禁用梯度计算")
print("=" * 60)

x = torch.randn(3, requires_grad=True)
print(f"x.requires_grad = {x.requires_grad}")

with torch.no_grad():
    y = x * 2
    print(f"在no_grad上下文中, y.requires_grad = {y.requires_grad}")

z = x * 2
print(f"在no_grad上下文外, z.requires_grad = {z.requires_grad}")

x_detached = x.detach()
print(f"\n使用detach()后, x_detached.requires_grad = {x_detached.requires_grad}")
```

## 2.8 综合练习：手动实现线性回归

```python
print("=" * 60)
print("2.8 使用autograd实现线性回归")
print("=" * 60)

torch.manual_seed(42)

X = torch.randn(100, 1)
true_w = 3.0
true_b = 2.0
y = true_w * X + true_b + torch.randn(100, 1) * 0.5

w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

learning_rate = 0.1
epochs = 50

print(f"真实参数: w = {true_w}, b = {true_b}")
print(f"初始参数: w = {w.item():.4f}, b = {b.item():.4f}")
print("\n开始训练...")

for epoch in range(epochs):
    y_pred = w * X + b
    
    loss = ((y_pred - y) ** 2).mean()
    
    if w.grad is not None:
        w.grad.zero_()
        b.grad.zero_()
    
    loss.backward()
    
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad
    
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}, w: {w.item():.4f}, b: {b.item():.4f}")

print(f"\n训练完成!")
print(f"学习到的参数: w = {w.item():.4f}, b = {b.item():.4f}")
print(f"真实参数: w = {true_w}, b = {true_b}")
```

## 2.9 链式法则验证

```python
print("=" * 60)
print("2.9 链式法则验证")
print("=" * 60)

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2
z = torch.sin(y)
w = z + 1

print(f"计算图: x -> y=x^2 -> z=sin(y) -> w=z+1")
print(f"x = {x.item()}")
print(f"y = {y.item():.4f}")
print(f"z = {z.item():.4f}")
print(f"w = {w.item():.4f}")

w.backward()

print(f"\ndw/dx (autograd) = {x.grad.item():.6f}")

manual_grad = 2 * x.item() * torch.cos(x.item() ** 2)
print(f"dw/dx (手动计算) = {manual_grad.item():.6f}")
print(f"两者一致: {torch.isclose(x.grad, manual_grad)}")
```
